# Text Preprocessing

```python
cols=['Unnamed: 2','Unnamed: 3','Unnamed: 4']
df=df.drop(cols,axis=1)
```

```python
df.head()
```

```python
df=df.rename(columns={
    'v1' : "types",
    'v2' : "sms"
})
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df['types'].unique()        
```

```python
df['types']=df['types'].map({
    "ham" : 0,
    "spam" : 1
})
```

```python
df.head()
```

```python
df['sms']=df['sms'].astype(str).str.lower()
```

```python
df.head()
```

```python
import string
def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['sms']=df['sms'].apply(remove_punc)
```

```python
df.head()
```

```python
def remove_num(txt):
    new=""
    for i in txt:
        if not i.isdigit():
            new=new+i
    return new
df['sms']=df['sms'].apply(remove_num)
```

```python
df.head()
```

```python
def remove_url(txt):
    new=""
    for i in txt:
        if i!="//" or i!="https" or i!=":":
            new=new+i
    return new
df['sms']=df['sms'].apply(remove_url)
```

```python
df
```

# Tokenization and Stopword Removal

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
stop_words=set(stopwords.words('english'))
```

```python
def remove(txt):
    words=word_tokenize(txt)
    cleaned=[]
    for i in words:
        if not i in stop_words:
            cleaned.append(i)
    return ' '.join(cleaned)
df['sms']=df['sms'].apply(remove)
```

```python
df.head()
```

# Model Selection

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['sms'], df['types'], test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(X_train)
X_test_bow = bow_vectorizer.transform(X_test)
```

### Multinomial Naive Bayes

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
mnb_model=MultinomialNB()
```

```python
mnb_model.fit(X_train_bow,y_train)
```

```python
y_pred=mnb_model.predict(X_test_bow)
```

```python
from sklearn.metrics import accuracy_score
acc_mnb=accuracy_score(y_test,y_pred)
print(acc_mnb)
```

### Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr_model=LogisticRegression()
lr_model.fit(X_train_bow,y_train)
```

```python
y_pred_lr=lr_model.predict(X_test_bow)
acc_lr=accuracy_score(y_test,y_pred_lr)
print(acc_lr)
```

### KNN

```python
from sklearn.neighbors import KNeighborsClassifier
knn_model=KNeighborsClassifier()
```

```python
knn_model.fit(X_train_bow,y_train)
y_pred_knn=knn_model.predict(X_test_bow)
acc_knn=accuracy_score(y_test,y_pred_knn)
print(acc_knn)
```

# Support Vector Machine

```python
from sklearn.svm import SVC
```

```python
svc_model=SVC()
svc_model.fit(X_train_bow,y_train)
```

```python
svc_model.fit(X_train_bow,y_train)
y_pred_svc=svc_model.predict(X_test_bow)
acc_svc=accuracy_score(y_test,y_pred_svc)
print(acc_svc)
```

### Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt_model=DecisionTreeClassifier()
dt_model.fit(X_train_bow,y_train)
```

```python
dt_model.fit(X_train_bow,y_train)
y_pred_dt=dt_model.predict(X_test_bow)
acc_dt=accuracy_score(y_test,y_pred_dt)
print(acc_dt)
```

### Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
rtc_model=RandomForestClassifier()
rtc_model.fit(X_train_bow,y_train)
```

```python
rtc_model.fit(X_train_bow,y_train)
y_pred_rtc=rtc_model.predict(X_test_bow)
acc_rtc=accuracy_score(y_test,y_pred_rtc)
print(acc_rtc)
```

# Results

```python
result = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Support Vector Machine",
        "Multinomial Naive Bayes",
        "KNN",
        "Decision Tree",
        "Random Forest"
    ],
    "Accuracy": [
        round(acc_lr, 3),
        round(acc_svc, 3),
        round(acc_mnb, 3),
        round(acc_knn, 3),
        round(acc_dt, 3),
        round(acc_rtc, 3)
    ]})
result
```

# Final Decision

#### Multinomial Naive Bayes achieved the highest accuracy of 98.0% among the tested models and was selected as the final model for SMS spam classification.
