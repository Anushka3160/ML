```python
columns=['mainroad','guestroom','basement','hotwaterheating','airconditioning','prefarea']
for col in columns:
    df[col]=df[col].map({
        'yes' : 1,
        'no' : 0
    })
```

```python
X=df.drop('price',axis=1)
y=df['price']
```

```python
X_one_encode=pd.get_dummies(X,columns=['furnishingstatus'],drop_first=True)
X_one_encode=X_one_encode.astype(int)
X_one_encode
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_one_encode, y, test_size=0.20, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
numeric_cols=['area','bedrooms','bathrooms','stories','parking']
X_train[numeric_cols]=scaler.fit_transform(X_train[numeric_cols])
X_test[numeric_cols]=scaler.fit_transform(X_test[numeric_cols])
```

```python
from sklearn.linear_model import LinearRegression
lr_model=LinearRegression()
```

```python
lr_model.fit(X_train,y_train)
```

```python
from sklearn.metrics import root_mean_squared_error, r2_score
```

```python
y_pred=lr_model.predict(X_test)
```

```python
r2=r2_score(y_test,y_pred)
r2
```

```python
rmse=root_mean_squared_error(y_test, y_pred)
rmse
```

```python
from sklearn.tree import DecisionTreeRegressor
dtr_model=DecisionTreeRegressor()
```

```python
dtr_model.fit(X_train,y_train)
```

```python
y_pred_dtr=dtr_model.predict(X_test)
```

```python
r2_dtr=r2_score(y_test,y_pred_dtr)
r2_dtr
```

```python
rmse_dtr=root_mean_squared_error(y_test,y_pred_dtr)
rmse_dtr
```

```python
from sklearn.ensemble import RandomForestRegressor
rfr_model=RandomForestRegressor(random_state=42)
```

```python
rfr_model.fit(X_train,y_train)
```

```python
y_pred_rfr=rfr_model.predict(X_test)
```

```python
r2_rfr=r2_score(y_test,y_pred_rfr)
r2_rfr
```

```python
rmse_rfr=root_mean_squared_error(y_test,y_pred_rfr)
rmse_rfr
```

```python
from sklearn.ensemble import GradientBoostingRegressor
gb_model=GradientBoostingRegressor()
```

```python
gb_model.fit(X_train,y_train)
```

```python
y_pred_gb=gb_model.predict(X_test)
```

```python
r2_gb=r2_score(y_test,y_pred_gb)
r2_gb
```

```python
rmse_gb=root_mean_squared_error(y_test,y_pred_gb)
rmse_gb
```
